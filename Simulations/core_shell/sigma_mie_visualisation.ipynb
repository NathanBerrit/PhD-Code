```python
def get_data_sigma_mie(fname):
    data = np.loadtxt(fname + '.dat', skiprows=1)

    Lambda = data[:,0]
    sigma_ext = data[:,4]
    sigma_sca = data[:,5]
    sigma_abs = data[:,6]

    res = {}
    res['Lambda'],res['s_ext'],res['s_sca'],res['s_abs'] = Lambda, sigma_ext, sigma_sca, sigma_abs
    return res
```

```python
dir_data = r'C:\Users\nberr\ENS_Lyon\M2\Stage\Core_shell'
os.chdir(dir_data)


Au_100 = get_data_sigma_mie('sphere_au_100nm')
Au_60 = get_data_sigma_mie('sphere_au_60nm')
Ag_40 = get_data_sigma_mie('sphere_ag_40nm')

fig = plt.figure(figsize=(10, 5))
plt.plot(Au_100['Lambda'][:500], Au_100['s_ext'][:500]*1e-3, color ='red', label='Sphère d\'Au, r = 100 nm')
plt.plot(Au_60['Lambda'][:500], Au_60['s_ext'][:500]*1e-3, color ='orange', label='Sphère d\'Au, r = 60 nm')
plt.plot(Ag_40['Lambda'][:500], Ag_40['s_ext'][:500]*1e-3, color ='black', label='Sphère d\'Ag, r = 40 nm')

plt.xlabel('$\lambda$ (nm)')
plt.ylabel(r'$\sigma_\mathrm{ext}$ ($10^{-3}$ nm$^2$)')
plt.tight_layout()
plt.legend(fontsize=15)
os.chdir(r'C:\Users\nberr\ENS_Lyon\M2\Stage\Rapport de stage\Figures\Diapo')
#plt.savefig('LSPR.pdf',bbox_inches='tight')
plt.show()
```
