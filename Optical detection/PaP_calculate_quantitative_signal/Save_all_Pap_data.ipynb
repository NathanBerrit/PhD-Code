## Save all data in an excel

```python
Dic_dataframe = {}

for key in All_traces :
    
    Trace = All_traces[key]

    lda_probe = str(Trace.lda_probe)



    delay = Trace.delay
    N = np.size(Trace.delay)

    Dic_dataframe[lda_probe + ' - delay'] = pd.Series(delay)
    Dic_dataframe[lda_probe + ' - DTT'] = pd.Series(Trace.DTT)
    func = globals()['ExpDec' + str(Trace.n_exp_thermic)]
    exp_key = str(Trace.n_exp_thermic) + ' exp'
    Trace.DTT_thermic_fit = func(Trace.delay,*Trace.thermal_fit[exp_key]['coeffs'])
    Dic_dataframe[lda_probe + ' - Thermic fit'] = pd.Series(Trace.DTT_thermic_fit)
    Dic_dataframe[lda_probe + ' - acoustic signal'] = pd.Series(Trace.DTT_acoustic)


    func_acoustic = globals()['Osc' + str(Trace.n_modes_acoustic)]
    plot_key = str(Trace.n_modes_acoustic) + ' modes'
    Trace.DTT_acoustic_fit = func_acoustic(Trace.delay,*Trace.acoustic_fit[plot_key]['coeffs'])


    Dic_dataframe[lda_probe + ' - delay tot'] = pd.Series(Trace.delay_tot)

    Dic_dataframe[lda_probe + ' - DTT tot'] = pd.Series(Trace.DTT_tot)


os.chdir(dir_save_data)
df = pd.DataFrame(Dic_dataframe)
df.to_csv('All PaP treated signals' + '.csv', index=False)
```
