## Fit the result from Christy 1980 for dependencies of scattering term

```python
T = np.array([78, 295, 425])
gamma_exp = np.array([0.7, 4.6, 7.3])
gamma_calc = np.array([1.37, 3.29, 4.76])

def affine(T, a, b) :
    return a*T+b




a_exp, b_exp = np.polyfit(T, gamma_exp, 1)
a_exp_lin = np.sum(T * gamma_exp) / np.sum(T**2)

a_calc, b_calc = np.polyfit(T, gamma_calc, 1)
a_calc_lin = np.sum(T * gamma_calc) / np.sum(T**2)

```

```python
fig, ax = plt.subplots()

ax.scatter(T, gamma_exp)
ax.plot(T, a_exp*T+b_exp)
ax.plot(T, a_exp_lin*T)
ax.set_title('Valeurs expérimentales')

ax.set_xlabel('T (K)')
ax.set_ylabel(r'$10^{13}$ $s^{-1}$')
ax.text(100, 5, 'fit affine a*x+b :\na = {:.3e} \nb = {:.3e}'.format(a_exp, b_exp))
ax.text(250, 1, 'fit linéaire (orange) :\na_lin = {:.3e}'.format(a_exp_lin))
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots()

ax.scatter(T, gamma_calc)
ax.plot(T, a_calc*T+b_calc)
ax.plot(T, a_calc_lin*T)
ax.set_title('Valeurs théoriques')

ax.set_xlabel('T (K)')
ax.set_ylabel(r'$10^{13}$ $s^{-1}$')
ax.text(100, 3, 'fit affine a*x+b :\na = {:.3e} \nb = {:.3e}'.format(a_calc, b_calc))
ax.text(250, 1, 'fit linéaire (orange) :\na_lin = {:.3e}'.format(a_calc_lin))
plt.tight_layout()
plt.show()
```
