```python
fig, ax = plt.subplots(figsize=(6,6))



Energy = np.linspace(1.5, 2.5, 1000)

x_probe = Energy[450]
delta_w = 0.07
tau = 800/3

delta_Ec = 0.01

#Lorentzian shape for the SPR
A = 1
E_c = 2
w = 0.3
y0=0


ax.set_xticks([])  # Remove x-axis ticks
ax.set_yticks([])  # Remove y-axis ticks
ax.set_xlabel(r'$\lambda$')
ax.set_ylabel(r'$\sigma_{ext}$')

ax.plot(Energy, Lorentz(Energy, E_c, w, A, y0),color='grey', label ='SPR at equilibrium')

ax.plot(Energy, Lorentz(Energy, E_c+ delta_Ec, w + delta_w, A, y0),color='blue', label = 'SPR after heating')

ax.annotate(text='', xy=(x_probe,Lorentz(x_probe, E_c, w, A, y0)), xytext=(x_probe,Lorentz(x_probe, E_c+ delta_Ec, w + delta_w, A, y0)),arrowprops=dict(arrowstyle='<-', color="#B97E00", linewidth=1.5))
ax.text(x_probe-0.15, 3, r'$\Delta \sigma_\mathrm{ext}$', color="#B97E00")

ax.axvline(E_c, color='black', linestyle='--',zorder=0)
ax.text(0.40, 0.05, r'$\lambda_\mathrm{SPR}$', transform=ax.transAxes)
fig.patch.set_facecolor( '#FF000000' )

plt.legend(fontsize=10)
plt.tight_layout()


os.chdir(dir_save_data)

plt.savefig('Simple_scheme_thermic_variation.svg')
```
